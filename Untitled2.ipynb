### Pembersihan Dataset (Data Cleaning)

Langkah-langkah yang akan dilakukan:
1. Menghapus baris duplikat.
2. Menangani nilai kosong pada kolom `Rating` (misalnya dengan nilai median atau mean, atau menghapusnya jika diperlukan).
3. Menangani nilai kosong pada kolom lainnya seperti `Type`, `Content Rating`, `Current Ver`, dan `Android Ver`.

```python
import pandas as pd

# Membaca dataset
df = pd.read_csv('/content/googleplaystore.csv')

# Memastikan nama kolom bersih dari spasi
df.columns = df.columns.str.strip()

print("Kolom yang terdeteksi di dalam file:", list(df.columns))

# 1. Menghapus baris duplikat
df_clean = df.drop_duplicates().copy()

# 2. Mengisi nilai kosong pada kolom 'Rating' jika kolom tersebut ada
if 'Rating' in df_clean.columns:
    median_rating = df_clean['Rating'].median()
    df_clean['Rating'] = df_clean['Rating'].fillna(median_rating)

# 3. Menghapus baris yang memiliki nilai kosong di kolom lainnya jika tersedia
columns_to_check = [col for col in ['Type', 'Content Rating', 'Current Ver', 'Android Ver'] if col in df_clean.columns]
if columns_to_check:
    df_clean = df_clean.dropna(subset=columns_to_check)

# Menampilkan hasil pembersihan
print("\n=== Setelah Pembersihan ===")
print("Ukuran dataset baru:", df_clean.shape)
print("Jumlah nilai kosong per kolom:\n", df_clean.isnull().sum())
print("Jumlah baris duplikat:", df_clean.duplicated().sum())
```

### Menghitung Korelasi antara Rating dan Reviews

Sebelum menghitung korelasi, kita perlu memastikan tipe data kedua kolom tersebut adalah numerik.

```python
import pandas as pd
import numpy as np

# Correct condition parenthesis
df_clean = df.copy()

# Remove row 10472 or filter invalid Rating > 5
df_clean = df_clean[(df_clean['Rating'] <= 5.0) | (df_clean['Rating'].isna())]

# Clean Reviews
df_clean['Reviews'] = pd.to_numeric(df_clean['Reviews'], errors='coerce')

# Clean Installs
df_clean['Installs_Clean'] = df_clean['Installs'].str.replace('+', '', regex=False).str.replace(',', '', regex=False)
df_clean['Installs_Clean'] = pd.to_numeric(df_clean['Installs_Clean'], errors='coerce')

# Clean Size to MB
def clean_size_mb(size_str):
    if pd.isna(size_str) or size_str == 'Varies with device':
        return np.nan
    size_str = str(size_str).strip()
    if size_str.endswith('M') or size_str.endswith('m'):
        return float(size_str[:-1])
    elif size_str.endswith('k') or size_str.endswith('K'):
        return float(size_str[:-1]) / 1024.0
    try:
        return float(size_str)
    except:
        return np.nan

df_clean['Size_MB'] = df_clean['Size'].apply(clean_size_mb)

print("Before imputation summary:")
print(df_clean[['Rating', 'Reviews', 'Installs_Clean', 'Size_MB']].isna().sum())

# Impute values
rating_median = round(df_clean['Rating'].median(), 2)
size_median = round(df_clean['Size_MB'].median(), 2)

df_clean['Rating'] = df_clean['Rating'].fillna(rating_median)
df_clean['Size_MB'] = df_clean['Size_MB'].fillna(size_median)

# Duplicates check
print("\nDuplicates count:", df_clean.duplicated().sum())

# Save both cleaned dataset (ready for ML/K-Means) and export CSV
output_df = df_clean[['Rating', 'Reviews', 'Installs_Clean', 'Size_MB']].rename(columns={
    'Installs_Clean': 'Installs',
    'Size_MB': 'Size_MB'
})

output_filename = 'googleplaystore_cleaned.csv'
output_df.to_csv(output_filename, index=False)

print("\nFinal Cleaned Data Head:")
print(output_df.head(10))
print("\nFinal Cleaned Data Summary:")
print(output_df.describe())
```
